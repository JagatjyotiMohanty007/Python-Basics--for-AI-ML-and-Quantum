```python
class student:
    def __init__(self,m1,m2):
        self.m1 = m1
        self.m2 = m2
        
    def __add__(self, other): ## WE DEFINE THE ADD METHOD . THIS CONCEPT IS CALLED OPERATOR OVERLOADING
        print(self.m1)
        print(self.m2)
        print(other.m1)
        print(other.m2)
        m1 = self.m1 + other.m1
        m2 = self.m2 + other.m2
        s3 = student(m1,m2)
        
        return s3   
 
s1 = student(60,70)
s2 = student(80,90)
s4 = student(50,40)

s3 = s4 + s1

print(s3.m1)
print(s3.m2)
```

```python
#if the see the above error we got error as > is not support so we need to define the > method (gt)
class student:
    def __init__(self,m1,m2):
        self.m1 = m1
        self.m2 = m2
    def __add__(self, other): ## WE DEFINE THE ADD METHOD . THIS CONCEPT IS CALLED OPERATOR OVERLOADING
        m1 = self.m1 + other.m1
        m2 = self.m2 + other.m2
        s3 = student(m1,m2)
        
        return s3  
    def __gt__(self,other):
        r1 = self.m1 + self.m2
        r2 = other.m1 + other.m2
        if r1 > r2:
            return True
        else:
            return False
 
s1 = student(60,70)
s2 = student(80,90)

s3 = s1 + s2 

if s1>s2:
    print('s1 pass')
else:
    print('s2 pass')
    
# we are not define it so how to check it 
```

```python
class A:
    def show(self):
        print('in A show')
class B:
    pass
      
s1 = A()
s1.show()
```
