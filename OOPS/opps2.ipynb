```python
class Student:
    school = 'NIT'      # Class (static) variable

    def __init__(self, m1, m2, m3):
        self.m1 = m1    # Instance variable
        self.m2 = m2
        self.m3 = m3

    def avg(self):
        return (self.m1 + self.m2 + self.m3) / 3

    @classmethod
    def info(cls):
        return cls.school

    @staticmethod
    def info1():
        print(Student.school)
        return  "This is a Student class."


s1 = Student(10, 20, 30)
s2 = Student(20, 30, 40)

s1.school = "IIT"   # Creates an instance variable for s1

print(s1.school)          # IIT
print(s2.school)          # NIT
print(Student.school)     # NIT

print(s1.avg())           # 20.0
print(Student.info())     # NIT
print(Student.info1())    # This is a Student class.
```

1. Class Method (@classmethod)

```python
class Student:
    school = "NIT"

    @classmethod
    def get_school(cls):
        return cls.school

print(Student.get_school())
```

```python
class Student:
    school = "NIT"

    @classmethod
    def change_school(cls, name):
        cls.school = name

Student.change_school("IIT")

print(Student.school)
```

# 2. Static Method (@staticmethod)

```python
class Student:
    school = "NIT"

    def __int__(self):
        pass

    def showschool(self):
        print(Student.school)
    

    @staticmethod
    def show(name):
        Student.school = name
        print(Student.school)

s1 = Student()
s2 = Student()

s1.showschool()
s1.show("123")
s2.showschool()
```

```python
class Student:
    school = "NIT"

    def __init__(self, name):
        self.name = name

    @classmethod
    def change_school(cls, new_name):
        cls.school = new_name

    @staticmethod
    def add(a, b):
        return a + b


s1 = Student("Jagatjyoti")

print(Student.school)

Student.change_school("IIT")

print(Student.school)

print(Student.add(10, 20))
```
