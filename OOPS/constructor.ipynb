Constructor in Inheritence

```python
class A(): 
    def __init__(self):
        print('in a init')     
    def feature1(self):         
        print('feature1 is hi')
    def feature2(self):
        print('feature2 is hi')
        
class B():
    def feature3(self):
        print('feature3 is hello')
    def feature4(self):
        print('feature4 is hello')

a1 = A()
b1 = B()
```

```python
class A(): 
    def __init__(self):
        print('in a init')     
    def feature1(self):         
        print('feature1 is hi')
    def feature2(self):
        print('feature2 is hi')
class B(A):
    def __init__(self):
        super().__init__()
        print('in b init')
    def feature3(self):
        print('feature3 is hello')
    def feature4(self):
        print('feature4 is hello')  
b1 = B()
```

```python
class A(): 
    def __init__(self):
        print('in the A init')
    def feature1(self):         # lets take example class A has 2 feature1 & feature 2
        print('feature1 is hi')
    def feature2(self):
        print('feature2 is hi')
class B():
    def __init__(self):
        print('in the B init')
    def feature3(self):         # lets take example class A has 2 feature1 & feature 2
        print('feature3 is hallo')
    def feature4(self):
        print('feature4 is hallo')
class C(A,B):
    def __init__(self):
        super().__init__()
        print('in the C init')                
c1 = C()
```
